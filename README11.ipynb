Project Summary: This data analysis workflow cleans raw, missing Kenyan county rainfall data—imputing incomplete monthly records for Machakos using calculated averages—and leverages 2D NumPy array operations to compute annual totals, peak wet months, and dry period frequencies.

How to Run: Clone the repository, install dependencies by running pip install numpy in your terminal, and execute all cells sequentially in Jupyter Notebook  to run the data cleaning pipeline and array operations.

Planting Advisory: Kisumu supports year-round rainfed crops with zero dry months, while high-rainfall counties (Nairobi, Nakuru, Mombasa) should plant ahead of their peak April–May rains, and semi-arid Machakos requires water harvesting or drought-tolerant crops to navigate its 6 dry months.
